```python
# Coding Challenge 
# Rmove Vowels

def remove_vowels(s):
    vowels = 'aeiouAEIOU'
    final = ""
    for letter in s:
        if letter not in vowels: 
            final += letter
    return final
```

```python
remove_vowels("give")
```

```python
# reverse String

def reverse_string(string):
    reverse = ""
    for s in string:
        reverse = s + reverse
    return reverse

# or another way
def reverse_str(string):
    return string[::-1]
```

```python
reverse_string("whatever")
```

```python
reverse_str("whatever")
```

### Lists

```python
elements = ['Apple', 'watermelon', 'pears', 'tomatos']
print(elements)
```

```python
print(elements[-1])
elements[0] == elements[-len(elements)]
```

```python
elements.append('crazy brains')    #add a new elements
elements.insert(0, 'Mango')
print(elements)
```

```python
del elements[2:4:2]   # similar to slicing (deleting)
print(elements)
```

```python
elements.pop() # removes last elements
elements.remove('Mango')     # removes the element named
print(elements)
```

```python
elements.sort()         # permanent sort()
print(sorted(elements)) # temporary sort --> doesnt edit the original list
print(elements)
```

```python
# Iteration

for e in elements:
    print(e.upper())

veggies = ['cucmber', 'carrot', 'lettuce', 'others']
print(sorted(elements + veggies))
```

#### Dictionaries

```python
about_person = { 'name':'Fred',
                'age' : 32,
                'is_single': True }
print(about_person)
print(about_person['age'])
```

```python
about_person['has_job'] = True    # adding to dict
about_person['age'] = 16
print(about_person)
```

```python
# removing
del about_person['name']
print(about_person)
```

```python
#iteration
for k,v in about_person.items():   # need the item() method to iterate
    print(k)
    print(type(v))    # inspects the type

about_person.items()     # another way to print
```

#### Sets

```python
# initialization
days = {'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'}
print(days)
days.add('Saturday')      # add more elements
print(days)
```

```python
# unique letters
name = set("chaithra")
print(name)         # prints only one of the letter if that letter is repeated
```

```python
# identify overlapping items
vowel = {'a','e','i','o','u'}
print(name.intersection(vowel))
```
