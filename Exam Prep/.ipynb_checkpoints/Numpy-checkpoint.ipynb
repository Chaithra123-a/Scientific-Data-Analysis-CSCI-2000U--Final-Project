```python
import numpy as np
import pandas as pd
```

```python
nums = np.array([2,4,6,8,9])
print(type(nums))
print(nums)
```

```python
# Vectorized operations
print(nums*2)
print(nums/2)
print(nums**2)    # power 2
print(nums)    # the above operations do not change any of the above
```

```python
num2 = np.array([2,4,'list'])
print(num2)          # prints everything as string, since the array has various types
```

```python
# 2x3 array
num3 = np.array([[1,2,3], [4,5,6]])
print(num3)
num3.dtype   # data type
```

```python
num3.ndim   # ndim is the no. of dimensions
```

```python
num3.shape
```

```python
num3.size
```

```python
# iteration  ----> since the dimention is 2 we have to iterate both rows and columns
for row in num3:
    for column in row:
        print(column, end="-")
    print()
```

```python
# flatten the array --> making it to 1 dimension
for e in num3.flat:
    print(e, end=' ')

num3.flat
```

##### Filling and Creating

```python
d = np.zeros(5)
print(d)
```

```python
#3x2 array containing only 1 value
d = np.full((3,2), 22)
print(d)
```

```python
# creating with range functions
np.arange(start=1, stop=20, step=5)   # ARANGE
```

```python
# evenly spaced/linearly spaced floats
np.linspace(start=1, stop=20, num=9)
```

```python
# reshape the array
np.linspace(start=1, stop=20, num=12).reshape(3,4)
```
