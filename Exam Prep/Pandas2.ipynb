```python
import numpy as np
import pandas as pd
```

```python
d = pd.Series ([4.5, 7.2, -5.3, 3.61], index=["d", "b", "a", "c"])
d
```

#### Reindexing

```python
d2 = d.reindex(["e", "a", "d", "c", "b"])
d2
```

```python
# coding challange, get the sorted version of d2
d2 = d.reindex(sorted(["e", "a", "d", "c", "b"]))
d2
```

#### Forward -filling using reindexing

```python
d = pd.Series(["apple", "pear", "banana"], index= [0, 2, 7])
d
print (d.reindex(np.arange (12)), "\n")    # adding more indexes
```

```python
print (d. reindex(np.arange (12), method="ffill"))   # to fill up all Nan
```

```python
print (d.reindex(np.arange (12), method="nearest"))  # near indicing
```

#### Reindexing Rows and Columns

```python
provinces = pd.DataFrame(np.arange (9).reshape((3, 3)), 
                           index= ["a", "c", "d"],  # rows
                           columns= ["Ontario" , "Quebec", "Alberta"])
print(provinces)
```

```python
# Row reindexing
p2 = provinces. reindex(index=["a", "b", "c", "d"])
p2
```

```python
p3 = provinces.reindex(columns=["Alberta", "Ontario", "Alberta"])
p3
```

#### Boolean Indexing

```python
grades = pd.DataFrame([[75, 45, 60], [100, 85, 92],
                      [55, 78, 0], [14, 88, 49]], 
                     columns=["Mark", "Kelly", "Idra"], 
                     index=["Exam1", "Exam2", "Exam3", "Exam4"])
grades
```

```python
grades.shape
```

```python
grades >= 50
```

```python
grades[grades<=50]
```

```python
grades[grades>=50]
```

```python
grades[(grades>=70) & (grades< 80)]
```

```python
# alternative way
idx = grades[(grades>=70) & (grades< 80)]
idx
```
