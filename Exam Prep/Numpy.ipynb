#### Operation and Calculations

```python
print(f"{num3*2}\n")
print(f"{num3**3}\n")
num3 + num3
```

```python
# broadcasting and some methods
grades = np.array([[75,59,60], [100, 85, 92],
                [55, 78, 93], [14,88,65]])
print(grades)
grades.shape
```

```python
print(f"Mean: {grades.mean()}")
print(f"Sum: {grades.sum()}")
print(f"Max: {grades.max()}")
print(f"std: {round(grades.std(),2)}")
```

#### Calculations by Rows or columns

```python
print(f"Mean: {grades.mean(axis = 0)}")    # gets average of each columns
```

```python
print(f"Mean: {grades.mean(axis = 1)}")    # gets average of each row
```

#### BroadCasting and rules

```python
print(grades/2)     # works since the digit 2 is stretched to the grade.shape()
```

```python
print(grades+grades)  # works cuz they have same shape doesn't have to be same list
```

#### Slicing and indexing

```python
print(grades)
print(grades[3,1])  # prints 1 specific needed no.
```

```python
print(grades[2,])  # prints 3rd row
print(grades[:,2]) # prints 3rd, column must have : in it or else it will be syntax
print(grades[0:2, 2]) # figure this out later
```

#### Shallow Copying

```python
marks = grades.view()
print(marks)
# if any edits are made to the grades, then marks is also changed
```

```python
marks +=2
print(grades)
print(marks)     # and yes if marks is changed, then so does grades like a pointer
```

#### Slice view()

```python
d = np.arange(1,6)
d_slice = d[:3]
print(d)
print(d_slice)
print(id(d))
print(id(d_slice))
```

```python
d += 10
print(d)
print(d_slice)
```

```python
d_slice -= 10
print(d)
print(d_slice)
```

#### Deep Copy

```python
d_copy = d.copy()
print(d_copy)
d_copy +=2         # creates an idividual duplicate, doesn't link both 
print(d_copy)
print(d)
```
