#### Writing CSV file

```python
import csv

# code to write initial stuff to file 
with open('employee.csv', mode='w', newline='') as emp_file:
    writer = csv.writer(emp_file)  
    writer.writerow(['ID','Name','Value'])  
    writer.writerow([134,'mini',10])
    writer.writerow([34,'geetha',9.5])
    writer.writerow([76,'Isha',7])
    writer.writerow([46,'kavri',9.5])
    writer.writerow([27,'Fassil',4])


# append something to the file:
with open('employee.csv', mode ='a', newline='') as emp_file:
    writer = csv.writer(emp_file)
    writer.writerow([12,'vennela',3])
```

#### Reading CSV file

```python
with open('employee.csv', mode ='r', newline='') as emp_file:
    reader = csv.reader(emp_file)
    for row in reader:
        id, name, value = row
        print(f"{id}\t{name}\t{value}")
```
