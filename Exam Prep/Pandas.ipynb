### Series 
similar to array but 1D and same type

```python
d = pd.Series([4,2,100,-35])
len(d)
```

```python
print(type(d))
print(d.dtype)
```

```python
print(d.array, "\n--------")
print(d.index)  
```

#### Series Index

```python
d1 = pd.Series([4,2,100,-35], index=["a","b","c","d"])
print(d1.array, "\n--------")
d1.index
d1['b']    # you give the position name to series by indexing it
```

```python
print(d1['a']+d1['c'], "\n")
print(d1[['a','b','c']], "\n")    # need 2 brackets
print(d1['b':'d'], "\n")
print(d1[['b','d']], "\n")
```

#### NumPy operations in Series

```python
d2 = d1.copy()
print(d2,"\n-----")
print(d2*2,"\n-----")
print('b' in d2)
```

#### Serires from Dictionaries

```python
prov_dict = {
    "Ontario": 7000,
    "Quebec": 45000,
    "British Columbia": 29500
}

prov = pd.Series(prov_dict)
print(prov,"\n-----")
```

```python
print(prov*2,"\n------")
prov['Ontario']
```

```python
prov['Ontario'] *2
```

```python
prov_codes = ["Ontario","Quebec","Alberta"]
prov2 = pd.Series(prov_dict, index=prov_codes)
print(prov2)
```

#### Index alignment

```python
prov + prov2
```

```python
prov.array + prov2.array
```

#### update index

```python
prov2.index = ["Ontario","Quebec","British Columbia"]
print(prov2)
prov + prov2
```
